## Gráfica de las curvas de las series de Taylor de varios órdenes

### Caso 1: 


$$f(x)=\frac{1}{25x^2+1}$$


$$
x_0=0
$$


Los polinomios de aproximación de Taylor son:

$$
P_0(x)=1
$$

$$
P_1(x)=1
$$

$$
P_2(x)=1-25x^2
$$

$$
P_3(x)=1-25x^2
$$

$$
P_4(x)=1-25x^2+625x^4
$$





```python
import sympy as sp

x = sp.Symbol('x')
f = 1/(1+25*x**2)

derivs = [f.diff(x, k) for k in range(5)]
valores_en_0 = [expr.subs(x, 0)for expr in derivs]

def taylor_serie(x0, grado):
    pol = 0
    for k in range(grado + 1):
        termino = valores_en_0[k] * (x-x0)**k / sp.factorial(k)
        pol += termino
    return pol

x0 = 0
max_ordem = 4

for n in range(max_ordem + 1):
    t = taylor_serie(x0,n)
    print(f"Grado {n} : {sp.simplify(t)}")
    
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_min = 0
x_max = 2
y_min = -0.1
y_max = 1.3
num_puntos = 500

x0 = 1

def f_arctan(x):
    return np.arctan(x)

def P0(x):
    return (np.pi / 4) * np.ones_like(x)

def P1(x):
    return (np.pi / 4) + (x - x0) / 2

def P2(x):
    return (np.pi / 4) + (x - x0) / 2 - ((x - x0)**2) / 4

def P3(x):
    return (
        (np.pi / 4)
        + (x - x0) / 2
        - ((x - x0)**2) / 4
        + ((x - x0)**3) / 12
    )

def P4(x):
    return (
        (np.pi / 4)
        + (x - x0) / 2
        - ((x - x0)**2) / 4
        + ((x - x0)**3) / 12
    )



x = np.linspace(x_min, x_max, num_puntos)

plt.figure(figsize=(10, 6))

plt.plot(x, f_arctan(x), 'k', label='arctan(x)', linewidth=2)
plt.plot(x, P0(x), '--', label='P0(x)')
plt.plot(x, P1(x), '--', label='P1(x)')
plt.plot(x, P2(x), '--', label='P2(x)')
plt.plot(x, P3(x), '--', label='P3(x)')
plt.plot(x, P4(x), '--', label='P4(x)')

plt.axvline(x=x0, color='gray', linestyle=':', label=f'x₀ = {x0}')
plt.scatter(x0, f_arctan(x0), color='red', zorder=5, label=f'Punto ({x0}, arctan({x0}))')

plt.title(f"Aproximaciones de Taylor de arctan(x) alrededor de x₀ = {x0}", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)

plt.grid(True, linestyle=':')
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.show()
```
