## ¿Qué es la complejidad computacional?

La complejidad computacional estudia la cantidad de recursos que necesita un algoritmo para resolver un problema.

Los recursos más importantes son:

- Tiempo de ejecución.
- Memoria utilizada.
- Número de operaciones.
- Tamaño de la entrada.

Normalmente se utiliza \(n\) para representar el tamaño de los datos.

La complejidad no busca medir segundos exactos, sino analizar cómo crece el costo de un algoritmo cuando aumenta \(n\).

---

## Notación Big O

Big O, escrita como \(O\), representa una cota superior del crecimiento de un algoritmo.

Por ejemplo:

$$
T(n)=3n^2+5n+8
$$

Para valores grandes de \(n\), el término dominante es \(n^2\), por lo que:

$$
T(n)=O(n^2)
$$

Big O se utiliza frecuentemente para describir el peor caso, pero no significa exclusivamente peor caso. También puede aplicarse al mejor caso o al caso promedio.

---

## Principales órdenes de complejidad

| Complejidad | Nombre | Ejemplo |
|---|---|---|
| \(O(1)\) | Constante | Acceder a una posición |
| \(O(\log n)\) | Logarítmica | Búsqueda binaria |
| \(O(n)\) | Lineal | Recorrer una lista |
| \(O(n\log n)\) | Lineal-logarítmica | Merge Sort |
| \(O(n^2)\) | Cuadrática | Dos ciclos anidados |
| \(O(2^n)\) | Exponencial | Probar subconjuntos |
| \(O(n!)\) | Factorial | Probar permutaciones |

El orden de crecimiento es:

$$
O(1)<O(\log n)<O(n)<O(n\log n)<O(n^2)<O(2^n)<O(n!)
$$

# Notaciones asintóticas

## Big O: \(O\)

Representa una cota superior.

$$
3n^2+4n+1=O(n^2)
$$

## Little o: \(o\)

Representa una cota superior estricta. Indica que una función crece más lentamente que otra.

$$
n=o(n^2)
$$

## Big Omega: \(\Omega\)

Representa una cota inferior.

$$
5n^2+3n=\Omega(n^2)
$$

## Big Theta: \(\Theta\)

Representa una cota ajustada, es decir, una función tiene el mismo orden de crecimiento superior e inferior.

$$
3n^2+5n+8=\Theta(n^2)
$$

| Notación | Significado |
|---|---|
| \(O\) | Cota superior |
| \(o\) | Cota superior estricta |
| \(\Omega\) | Cota inferior |
| \(\Theta\) | Cota ajustada |

---

# Mejor caso, peor caso y caso promedio

En una búsqueda lineal:

- **Mejor caso:** el elemento está al inicio.

$$
T_{\text{mejor}}(n)=\Theta(1)
$$

- **Peor caso:** el elemento está al final o no existe.

$$
T_{\text{peor}}(n)=\Theta(n)
$$

- **Caso promedio:** se revisa aproximadamente la mitad de la lista.

$$
T_{\text{promedio}}(n)=\Theta(n)
$$

# Problema P vs NP

El problema P vs NP pregunta:

> Si una solución puede verificarse rápidamente, ¿también puede encontrarse rápidamente?

## Clase P

P contiene problemas que pueden resolverse en tiempo polinomial.

Ejemplos:

- Búsqueda binaria.
- Ordenamiento.
- Máximo común divisor.
- Camino más corto.
- Sistemas de ecuaciones lineales.

## Clase NP

NP contiene problemas cuyas soluciones pueden verificarse en tiempo polinomial.

Esto no significa necesariamente que puedan resolverse rápidamente.

Todo problema de P también pertenece a NP:

$$
P\subseteq NP
$$

---

## Ejemplo de problema en NP

En el problema de suma de subconjuntos se busca una combinación que alcance un valor determinado.

Por ejemplo:

```text
Conjunto: {3, 7, 10, 12}
Objetivo: 19
```

Una solución es:

```text
7 + 12 = 19
```

Encontrar la combinación puede ser difícil, pero verificarla es sencillo.

---

## NP-completo y NP-hard

- **NP-completo:** problemas más difíciles dentro de NP.
- **NP-hard:** problemas al menos tan difíciles como los NP-completos, aunque no necesariamente pertenezcan a NP.

Ejemplos de problemas NP-completos:

- SAT.
- Suma de subconjuntos.
- Coloreo de grafos.
- Camino hamiltoniano.
- Problema de la mochila en su versión de decisión.

Si se encontrara un algoritmo polinomial para un problema NP-completo, entonces:

$$
P=NP
$$

---

## ¿P es igual a NP?

Actualmente no se ha demostrado si:

$$
P=NP
$$

o si:

$$
P\neq NP
$$

La mayoría de los investigadores considera probable que:

$$
P\neq NP
$$

Este problema es importante porque afecta áreas como:

- Criptografía.
- Inteligencia artificial.
- Optimización.
- Logística.
- Planificación.

---

## Conclusión

La complejidad computacional permite analizar cómo aumenta el tiempo y la memoria requeridos por un algoritmo.

Big O representa una cota superior, Big Omega una cota inferior y Big Theta una cota ajustada.

El problema P vs NP pregunta si todos los problemas cuyas soluciones pueden verificarse rápidamente también pueden resolverse rápidamente. Hasta el momento, este problema continúa sin una solución definitiva.
