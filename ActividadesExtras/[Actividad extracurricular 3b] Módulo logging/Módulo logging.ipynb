## 1. Investigación sobre el módulo `logging`

El módulo logging de Python permite registrar los eventos que ocurren durante
la ejecución de un programa.

A diferencia de print(), logging permite:

- Clasificar los mensajes por nivel de importancia.
- Guardar los mensajes en archivos.
- Mostrar información adicional como fecha, hora, archivo y línea.
- Enviar los mensajes a diferentes destinos.
- Filtrar qué mensajes se muestran o almacenan.

Los niveles principales son:

| Nivel | Descripción |
|---|---|
| DEBUG | Información detallada utilizada para depuración. |
| INFO | Indica que el programa funciona correctamente. |
| WARNING | Advierte sobre una situación inesperada. |
| ERROR | Indica que ocurrió un error. |
| CRITICAL | Representa un error grave que puede detener el programa. |

```python
import logging
import requests
import getpass

from colorlog import ColoredFormatter
```

```python
TELEGRAM_CHAT_ID = "1702550693"
```

```python
url = f"https://api.telegram.org/bot{TELEGRAM_TOKEN}/sendMessage"

datos = {
    "chat_id": TELEGRAM_CHAT_ID,
    "text": "Prueba"
}

try:
    respuesta = requests.post(
        url,
        data=datos,
        timeout=10
    )

    respuesta.raise_for_status()

    resultado = respuesta.json()

    if resultado.get("ok"):
        print("Mensaje enviado correctamente.")
    else:
        print("Telegram respondió con un problema:")
        print(resultado)

except requests.RequestException as error:
    print("No fue posible enviar el mensaje.")
    print("Detalle:", error)
```

```python
class TelegramHandler(logging.Handler):
    """
    Handler personalizado que envía mensajes de logging a Telegram.
    """

    def emit(self, record):
        try:
            mensaje = self.format(record)

            url = (
                f"https://api.telegram.org/"
                f"bot{TELEGRAM_TOKEN}/sendMessage"
            )

            datos = {
                "chat_id": TELEGRAM_CHAT_ID,
                "text": mensaje
            }

            respuesta = requests.post(
                url,
                data=datos,
                timeout=10
            )

            respuesta.raise_for_status()

        except requests.RequestException as error:
            print(
                "Error enviando el registro a Telegram:",
                error
            )
```

```python
logger = logging.getLogger("LoggerActividad")

logger.setLevel(logging.DEBUG)

logger.propagate = False
logger.handlers.clear()

print("Logger principal creado correctamente.")
```

```python
formato_consola = (
    "%(log_color)s"
    "[%(levelname)s] "
    "%(asctime)s - "
    "%(filename)s - "
    "Línea %(lineno)d - "
    "%(message)s"
)

console_handler = logging.StreamHandler()

console_handler.setLevel(logging.DEBUG)

console_formatter = ColoredFormatter(
    formato_consola,
    datefmt="%Y-%m-%d %H:%M:%S",
    log_colors={
        "DEBUG": "cyan",
        "INFO": "green",
        "WARNING": "yellow",
        "ERROR": "red",
        "CRITICAL": "bold_red"
    }
)

console_handler.setFormatter(console_formatter)

logger.addHandler(console_handler)

print("Logging en consola configurado.")
```

```python
file_handler = logging.FileHandler(
    "aplicacion.log",
    mode="a",
    encoding="utf-8"
)

file_handler.setLevel(logging.INFO)

file_formatter = logging.Formatter(
    (
        "[%(levelname)s] "
        "%(asctime)s - "
        "%(filename)s - "
        "Línea %(lineno)d - "
        "%(message)s"
    ),
    datefmt="%Y-%m-%d %H:%M:%S"
)

file_handler.setFormatter(file_formatter)

logger.addHandler(file_handler)

print("Logging en archivo configurado.")
```

```python
telegram_handler = TelegramHandler()

telegram_handler.setLevel(logging.ERROR)

telegram_formatter = logging.Formatter(
    (
        "🚨 REGISTRO DE PYTHON\n\n"
        "Nivel: %(levelname)s\n"
        "Fecha: %(asctime)s\n"
        "Archivo: %(filename)s\n"
        "Línea: %(lineno)d\n"
        "Mensaje: %(message)s"
    ),
    datefmt="%Y-%m-%d %H:%M:%S"
)

telegram_handler.setFormatter(
    telegram_formatter
)

logger.addHandler(telegram_handler)

print("Logging en Telegram configurado.")
```
