### ¿Qué es Prompt Engineering?

El prompt engineering, o ingeniería de prompts, consiste en diseñar instrucciones claras y estructuradas para comunicarse con modelos de inteligencia artificial.

El objetivo es conseguir respuestas más precisas, relevantes y adaptadas a una necesidad específica. Un buen prompt no solo contiene una pregunta, sino también el contexto, el objetivo, las restricciones y la forma en la que se espera recibir la respuesta.

### 1. Pensar el prompt como una instrucción estructurada

Un prompt no debe verse únicamente como una pregunta sencilla, sino como un conjunto de instrucciones que guía el comportamiento del modelo.

Mientras más clara sea la tarea, menor será la posibilidad de que la inteligencia artificial interprete incorrectamente lo solicitado.

Un prompt estructurado puede incluir:

- La tarea que debe realizarse.
- El contexto necesario.
- Las restricciones.
- El formato esperado.
- Los criterios que debe cumplir la respuesta.


### 2. Definir un rol o persona

Una recomendación importante es indicar qué papel debe asumir la inteligencia artificial.

El rol ayuda a establecer:

- El nivel de conocimiento.
- El tipo de vocabulario.
- El tono de la respuesta.
- La profundidad de la explicación.
- El enfoque que debe utilizarse.


### 3. Proporcionar suficiente contexto

La inteligencia artificial necesita información suficiente para comprender correctamente la situación.

Cuando el prompt no contiene contexto, el modelo puede responder de manera demasiado general, hacer suposiciones incorrectas o incluir información que no corresponde a la necesidad del usuario.

El contexto puede incluir:

- El tema que se está estudiando.
- El nivel de conocimiento del usuario.
- Los datos disponibles.
- El objetivo final.
- Los métodos que deben utilizarse.
- Las condiciones que no deben modificarse.

 

### 4. Especificar el formato y las restricciones de salida

También es importante indicar cómo debe presentarse la respuesta.

Se pueden establecer restricciones relacionadas con:

- Número de palabras.
- Cantidad de ejemplos.
- Uso de tablas o listas.
- Nivel técnico.
- Idioma.
- Tono.
- Estructura.
- Número de pasos.
- Cantidad de decimales.
- Elementos que no deben incluirse.



### 5. Incluir ejemplos y mejorar el prompt mediante pruebas

Cuando una tarea puede interpretarse de diferentes formas, es útil proporcionar uno o más ejemplos del resultado esperado.

Esta técnica se conoce como *few-shot prompting*. Los ejemplos permiten mostrar al modelo el patrón que debe seguir.

Los ejemplos ayudan a que la inteligencia artificial entienda mejor la estructura y el criterio de clasificación.
Además, un prompt no siempre queda perfecto en el primer intento. Es recomendable revisar la respuesta y mejorar progresivamente la instrucción.

El proceso puede realizarse así:

1. Escribir una primera versión del prompt.
2. Analizar la respuesta obtenida.
3. Identificar información incorrecta, faltante o innecesaria.
4. Agregar contexto o restricciones.
5. Probar nuevamente.
6. Comparar los resultados.
