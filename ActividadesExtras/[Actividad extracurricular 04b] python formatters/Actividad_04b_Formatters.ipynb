## Black

Black es un formatter “opinionado”. Esto significa que aplica un estilo definido y permite pocas configuraciones.

Su objetivo es que todos los archivos Python tengan una apariencia uniforme.

```python
import shutil

shutil.copy("archivo_original.py", "archivo_black.py")
!black archivo_black.py
```

```python
with open("archivo_black.py", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

## autopep8

autopep8 corrige automáticamente problemas de estilo siguiendo las recomendaciones de PEP 8.

PEP 8 es la guía oficial de estilo para código Python.

```python
shutil.copy("archivo_original.py", "archivo_autopep8.py")
!autopep8 --in-place archivo_autopep8.py
```

```python
with open("archivo_autopep8.py", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

## YAPF

YAPF significa “Yet Another Python Formatter”.

Fue desarrollado originalmente por Google y permite configurar distintos estilos de formato.

```python
shutil.copy("archivo_original.py", "archivo_yapf.py")
!yapf -i archivo_yapf.py
```

```python
with open("archivo_yapf.py", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

## Ruff

Ruff es una herramienta moderna y muy rápida.

Puede utilizarse como formatter y también como linter, es decir, puede detectar ciertos errores y problemas de estilo.

```python
shutil.copy("archivo_original.py", "archivo_ruff.py")
!ruff format archivo_ruff.py
```

```python
with open("archivo_ruff.py", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

## Comparación de los formatters

| Formatter | Característica principal |
|---|---|
| Black | Aplica un estilo uniforme y estricto |
| autopep8 | Corrige el código siguiendo PEP 8 |
| YAPF | Permite mayor configuración |
| Ruff | Es rápido y también funciona como linter |
