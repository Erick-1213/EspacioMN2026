```python
tokens = list(range(1_000_000))
```

```python
print(tokens[:10])
print(tokens[-10:])
```

Vector de Precios

```python
precio = []
for t in tokens:
    if t <= 272_000:
        precio.append(15 * t + 8 ) 
    else:
        precio.append(10 * t - 4)
```

```python
precio = []

precio = []
for t in tokens:
    if t <= 272_000:
        precio.append((30 / 1_000_000) * t)
    else:
        costo_tramo1 = (30 / 1_000_000) * 272_000          
        costo_tramo2 = (45 / 1_000_000) * (t - 272_000)    
        precio.append(costo_tramo1 + costo_tramo2)
```

```python
import matplotlib.pyplot as plt

plt.plot(tokens, precio)
plt.xlabel('Tokens')
plt.ylabel('Precio')
plt.title('Precio vs Tokens ')
plt.grid(alpha=0.35)

plt.axvline(x=272_000, linestyle='--', color='gray')
plt.scatter(272_000, 8.16, color='black', zorder=3)

plt.tight_layout()
plt.show()
```

```python
from scipy.optimize import bisect
import numpy as np
import matplotlib.pyplot as plt


def f(x):
    return (1/4) * (x**3 + 3*x**2 - 6*x - 8)


rangos = [[-5, 3], [-4.7, 2.5], [-4.5, 2.7], [-3, -2]]

print("Rango\t\tf(a)\tf(b)\t¿Cambio?\tRaíz")
print("-" * 60)

for a, b in rangos:
    fa = f(a)
    fb = f(b)
    cambio = "Sí" if fa * fb < 0 else "No"
    
    if fa * fb < 0:
        raiz = bisect(f, a, b)
        print(f"{[a, b]}\t{fa:.3f}\t{fb:.3f}\t{cambio}\t\tx = {raiz:.6f}")
    else:
        print(f"{[a, b]}\t{fa:.3f}\t{fb:.3f}\t{cambio}\t\t—")


x = np.linspace(-5, 3, 1000)
y = f(x)

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=1)
plt.grid(True)
plt.title('f(x) = 1/4(x³ + 3x² - 6x - 8)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
```
