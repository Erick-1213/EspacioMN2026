```python
import matplotlib.pyplot as plt

xs = [
    3.38,
    0.35,
    2.07,
    -0.45,
    -0.55,
    -1.06,
    -2.77,
    3.08,
    -3.98,
    -5.,
    -3.18,
    -1.96,
    2.37,
    -1.66,
    4.09
]

ys = [
    -0.04,
    1.28,
    0.65,
    2.18,
    1.70,
    1.96,
    3.36,
    0.18,
    3.35,
    4.16,
    2.95,
    2.34,
    0.38,
    1.75,
    -1.06
]
```

```python
plt.figure(figsize=(8,6))
plt.scatter(xs, ys, color='red')

plt.grid(True, linestyle='--', alpha=0.6)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gráfica de puntos')
plt.show()
```

```python
x = np.array(xs)
y = np.array(ys)

n = len(x)

m = (n*np.sum(x*y) - np.sum(x)*np.sum(y)) / (n*np.sum(x**2) - (np.sum(x))**2)

b = (np.sum(y) - m*np.sum(x)) / n

print("m =", m)
print("b =", b)

print(f"La recta es: y = {m:.4f}x + {b:.4f}")
```

```python
m, b = -0.5032325854363767, 1.4991976183166864

xs0 = np.linspace(-6, 40, 400)

def s_0(x):
    return m * x + b

plt.figure(figsize=(15, 6))

for xi, yi in zip(xs, ys):
    plt.scatter([xi], [yi], color='red')

plt.plot(xs0, s_0(xs0), label='s_0(x)', color='blue')

plt.grid(True, linestyle='--', alpha=0.6)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Mínimos cuadrados')
plt.legend()
plt.xlim(-6, 5)
plt.ylim(-2, 5)
plt.show()
```

```python
x = np.array(xs)
y = np.array(ys)

n = len(x)

A = np.array([
    [np.sum(x**4), np.sum(x**3), np.sum(x**2)],
    [np.sum(x**3), np.sum(x**2), np.sum(x)],
    [np.sum(x**2), np.sum(x), n]
])

B = np.array([
    np.sum((x**2) * y),
    np.sum(x * y),
    np.sum(y)
])

a, b, c = np.linalg.solve(A, B)

print("a =", a)
print("b =", b)
print("c =", c)

print(f"La función cuadrática es:")
print(f"y = {a:.4f}x² + {b:.4f}x + {c:.4f}")
```

```python
A = np.array([
    [np.sum(x**6), np.sum(x**5), np.sum(x**4), np.sum(x**3)],
    [np.sum(x**5), np.sum(x**4), np.sum(x**3), np.sum(x**2)],
    [np.sum(x**4), np.sum(x**3), np.sum(x**2), np.sum(x)],
    [np.sum(x**3), np.sum(x**2), np.sum(x), n]
])

B = np.array([
    np.sum((x**3) * y),
    np.sum((x**2) * y),
    np.sum(x * y),
    np.sum(y)
])

a, b, c, d = np.linalg.solve(A, B)

print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)

print(f"La función cúbica es:")
print(f"y = {a:.4f}x³ + {b:.4f}x² + {c:.4f}x + {d:.4f}")
```

# B.)

```python
x = np.array(xs)
y = np.array(ys)
n = len(x)

A_lineal = np.array([
    [np.sum(x**2), np.sum(x)],
    [np.sum(x), n]
])

B_lineal = np.array([
    np.sum(x*y),
    np.sum(y)
])

m, b_lineal = np.linalg.solve(A_lineal, B_lineal)

def s_lineal(x):
    return m*x + b_lineal


A_cuadratica = np.array([
    [np.sum(x**4), np.sum(x**3), np.sum(x**2)],
    [np.sum(x**3), np.sum(x**2), np.sum(x)],
    [np.sum(x**2), np.sum(x), n]
])

B_cuadratica = np.array([
    np.sum((x**2)*y),
    np.sum(x*y),
    np.sum(y)
])

a2, b2, c2 = np.linalg.solve(A_cuadratica, B_cuadratica)

def s_cuadratica(x):
    return a2*x**2 + b2*x + c2


A_cubica = np.array([
    [np.sum(x**6), np.sum(x**5), np.sum(x**4), np.sum(x**3)],
    [np.sum(x**5), np.sum(x**4), np.sum(x**3), np.sum(x**2)],
    [np.sum(x**4), np.sum(x**3), np.sum(x**2), np.sum(x)],
    [np.sum(x**3), np.sum(x**2), np.sum(x), n]
])

B_cubica = np.array([
    np.sum((x**3)*y),
    np.sum((x**2)*y),
    np.sum(x*y),
    np.sum(y)
])

a3, b3, c3, d3 = np.linalg.solve(A_cubica, B_cubica)

def s_cubica(x):
    return a3*x**3 + b3*x**2 + c3*x + d3



print("FUNCIÓN LINEAL")
print(f"y = {m:.6f}x + {b_lineal:.6f}")

print("\nFUNCIÓN CUADRÁTICA")
print(f"y = {a2:.6f}x² + {b2:.6f}x + {c2:.6f}")

print("\nFUNCIÓN CÚBICA")
print(f"y = {a3:.6f}x³ + {b3:.6f}x² + {c3:.6f}x + {d3:.6f}")



modelos = {
    "Lineal": s_lineal,
    "Cuadrática": s_cuadratica,
    "Cúbica": s_cubica
}

resultados = {}

for nombre, funcion in modelos.items():
    y_aprox = funcion(x)
    error_abs = np.abs(y - y_aprox)

    error_promedio = np.mean(error_abs)
    desviacion = np.std(error_abs)

    indice_max = np.argmax(error_abs)
    indice_min = np.argmin(error_abs)

    resultados[nombre] = { "error_abs": error_abs, "error_promedio": error_promedio, "desviacion": desviacion, "indice_max": indice_max, "indice_min": indice_min}

    print("\n=================================")
    print(nombre)
    print("=================================")
    print(f"Error absoluto promedio = {error_promedio:.6f}")
    print(f"Desviación estándar = {desviacion:.6f}")

    print("\nPunto con error máximo:")
    print(f"x = {x[indice_max]}, y = {y[indice_max]}")
    print(f"Error máximo = {error_abs[indice_max]:.6f}")

    print("\nPunto con error mínimo:")
    print(f"x = {x[indice_min]}, y = {y[indice_min]}")
    print(f"Error mínimo = {error_abs[indice_min]:.6f}")


xs0 = np.linspace(min(x)-1, max(x)+1, 500)

plt.figure(figsize=(15, 6))

plt.scatter(x, y, color='red',s = 70, label='Puntos originales')

plt.plot(xs0, s_lineal(xs0), color='green', label='Lineal')
plt.plot(xs0, s_cuadratica(xs0), color='blue', label='Cuadrática')
plt.plot(xs0, s_cubica(xs0), color='yellow', label='Cúbica')

colores = { "Lineal": "green", "Cuadrática": "blue", "Cúbica": "yellow"}

for nombre, funcion in modelos.items():
    indice_max = resultados[nombre]["indice_max"]
    indice_min = resultados[nombre]["indice_min"]

    plt.scatter( x[indice_max], y[indice_max], color=colores[nombre], marker='X', s=40, label=f'Error máximo {nombre}')

    plt.plot([x[indice_max], x[indice_max]], [y[indice_max], funcion(x[indice_max])], color=colores[nombre], linestyle='--')

    plt.scatter( x[indice_min], y[indice_min], color=colores[nombre], marker='D', s=40, label=f'Error mínimo {nombre}')

plt.grid(True, linestyle='--', alpha=0.6)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Mínimos cuadrados: lineal, cuadrática y cúbica')
plt.legend()
plt.xlim(min(x)-1, max(x)+1)
plt.ylim(min(y)-1, max(y)+1)
plt.show()
```
