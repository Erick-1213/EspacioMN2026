Grafique las curvas de las series de Taylor de varios órdenes para los siguientes casos:

- $\cos(x), \ x_0 = 0$

La función principal es:

$$
f(x)=\cos(x)
$$

Los polinomios de aproximación de Taylor son:

$$
P_0(x)=\cos(x_0)
$$

$$
P_1(x)=\cos(x_0)-\sin(x_0)(x-x_0)
$$

$$
P_2(x)=\cos(x_0)-\sin(x_0)(x-x_0)-\frac{\cos(x_0)}{2}(x-x_0)^2
$$

$$
P_3(x)=\cos(x_0)-\sin(x_0)(x-x_0)-\frac{\cos(x_0)}{2}(x-x_0)^2+\frac{\sin(x_0)}{6}(x-x_0)^3
$$

$$
P_4(x)=\cos(x_0)-\sin(x_0)(x-x_0)-\frac{\cos(x_0)}{2}(x-x_0)^2+\frac{\sin(x_0)}{6}(x-x_0)^3+\frac{\cos(x_0)}{24}(x-x_0)^4
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Límites de la gráfica
x_min = -10
x_max = 10  
y_min = -2
y_max = 2
num_puntos = 500

x0 = 0
   

def f_cos(x):
    return np.cos(x)

def P0(x):
    return np.cos(x0) * np.ones_like(x)

def P1(x):
    return np.cos(x0) - np.sin(x0) * (x - x0)

def P2(x):
    return np.cos(x0) - np.sin(x0) * (x - x0) - (np.cos(x0) / 2) * (x - x0)**2

def P3(x):
    return (
        np.cos(x0)
        - np.sin(x0) * (x - x0)
        - (np.cos(x0) / 2) * (x - x0)**2
        + (np.sin(x0) / 6) * (x - x0)**3
    )
 
def P4(x):
    return (
        np.cos(x0)
        - np.sin(x0) * (x - x0)
        - (np.cos(x0) / 2) * (x - x0)**2
        + (np.sin(x0) / 6) * (x - x0)**3
        + (np.cos(x0) / 24) * (x - x0)**4
    )

x = np.linspace(x_min, x_max, num_puntos)


plt.figure(figsize=(10, 6))

plt.plot(x, f_cos(x), 'k', label='cos(x)', linewidth=2)
plt.plot(x, P0(x), '--', label='P0(x)')
plt.plot(x, P1(x), '--', label='P1(x)')
plt.plot(x, P2(x), '--', label='P2(x)')
plt.plot(x, P3(x), '--', label='P3(x)')
plt.plot(x, P4(x), '--', label='P4(x)')         

plt.axvline(x=x0, color='gray', linestyle=':', label=f'x₀ = {x0}')
plt.scatter(x0, np.cos(x0), color='red', zorder=5, label=f'Punto ({x0}, cos({x0}))')

plt.title(f"Aproximaciones de Taylor de cos(x) alrededor de x₀ = {x0}", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)

plt.grid(True, linestyle=':')
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.show()
```

- $\dfrac{1}{1-x}$

La función principal es:

$$
f(x)=\frac{1}{1-x}
$$

Los polinomios de aproximación son:

$$
P_0(x)=\frac{1}{1-x_0}
$$

$$
P_1(x)=\frac{1}{1-x_0}+\frac{x-x_0}{(1-x_0)^2}
$$

$$
P_2(x)=\frac{1}{1-x_0}+\frac{x-x_0}{(1-x_0)^2}+\frac{(x-x_0)^2}{(1-x_0)^3}
$$

$$
P_3(x)=\frac{1}{1-x_0}+\frac{x-x_0}{(1-x_0)^2}+\frac{(x-x_0)^2}{(1-x_0)^3}+\frac{(x-x_0)^3}{(1-x_0)^4}
$$

$$
p_4(x)=\frac{1}{1-x_0}+\frac{x-x_0}{(1-x_0)^2}+\frac{(x-x_0)^2}{(1-x_0)^3}+\frac{(x-x_0)^3}{(1-x_0)^4+}+\frac{(x-x_0)^4}{(1-x_0)^5}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

x_min = -10
x_max = 10  
y_min = -2
y_max = 2
num_puntos = 500

x0 = 0   

def f(x):
    return 1 / (1 - x)

def P0(x):
    return (1 / (1 - x0)) * np.ones_like(x)

def P1(x):
    return 1/(1 - x0) + (x - x0)/(1 - x0)**2

def P2(x):
    return 1/(1 - x0) + (x - x0)/(1 - x0)**2 + (x - x0)**2/(1 - x0)**3

def P3(x):
    return (
        1/(1 - x0)
        + (x - x0)/(1 - x0)**2
        + (x - x0)**2/(1 - x0)**3
        + (x - x0)**3/(1 - x0)**4
    )
def P4(x):
    return (
        1/(1 - x0)
        + (x - x0)/(1 - x0)**2
        + (x - x0)**2/(1 - x0)**3
        + (x - x0)**3/(1 - x0)**4
        + (x - x0)**4/(1 - x0)**5
    )

x = np.linspace(x_min, x_max, num_puntos)


f_values = f(x)
f_values[np.abs(x - 1) < 0.05] = np.nan

plt.figure(figsize=(10, 6))

plt.plot(x, f_values, 'k', label=r'$f(x)=\frac{1}{1-x}$', linewidth=2)
plt.plot(x, P0(x), '--', label='P0(x)')
plt.plot(x, P1(x), '--', label='P1(x)')
plt.plot(x, P2(x), '--', label='P2(x)')
plt.plot(x, P3(x), '--', label='P3(x)')
plt.plot(x, P4(x), '--', label='P4(x)')

plt.axvline(x=x0, color='gray', linestyle=':', label=f'x₀ = {x0}')
plt.scatter(x0, f(x0), color='red', zorder=5, label=f'Punto ({x0}, f({x0}))')

plt.axvline(x=1, color='red', linestyle=':', label='Asíntota x = 1')

plt.title(f"Aproximaciones de Taylor de f(x) alrededor de x₀ = {x0}", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)

plt.grid(True, linestyle=':')
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.show()
```

-  $\ln(x), \ x_0 = 1$

La función principal es:

$$
f(x)=\ln(x)
$$

Los polinomios de aproximación de Taylor son:

$$
P_0(x)=\ln(x_0)
$$

$$
P_1(x)=\ln(x_0)+\frac{x-x_0}{x_0}
$$

$$
P_2(x)=\ln(x_0)+\frac{x-x_0}{x_0}-\frac{(x-x_0)^2}{2x_0^2}
$$

$$
P_3(x)=\ln(x_0)+\frac{x-x_0}{x_0}-\frac{(x-x_0)^2}{2x_0^2}+\frac{(x-x_0)^3}{3x_0^3}
$$

$$
P_4(x)=\ln(x_0)+\frac{x-x_0}{x_0}-\frac{(x-x_0)^2}{2x_0^2}+\frac{(x-x_0)^3}{3x_0^3}-\frac{(x-x_0)^4}{4x_0^4}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

x_min = 0.1
x_max = 3
y_min = -3
y_max = 2
num_puntos = 500

x0 = 1

def f(x):
    return np.log(x)

def P0(x):
    return np.zeros_like(x)

def P1(x):
    return x - x0

def P2(x):
    return (x - x0) - ((x - x0)**2) / 2

def P3(x):
    return (x - x0) - ((x - x0)**2) / 2 + ((x - x0)**3) / 3

x = np.linspace(x_min, x_max, num_puntos)

plt.figure(figsize=(10, 6))

plt.plot(x, f(x), 'k', label=r'$f(x)=\ln(x)$', linewidth=2)
plt.plot(x, P0(x), '--', label='P0(x)')
plt.plot(x, P1(x), '--', label='P1(x)')
plt.plot(x, P2(x), '--', label='P2(x)')
plt.plot(x, P3(x), '--', label='P3(x)')

plt.axvline(x=x0, color='gray', linestyle=':', label=f'x₀ = {x0}')
plt.scatter(x0, f(x0), color='red', zorder=5, label=f'Punto ({x0}, ln({x0}))')

plt.title(f"Aproximaciones de Taylor de ln(x) alrededor de x₀ = {x0}", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)

plt.grid(True, linestyle=':')
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.show()
```

Encuentre el polinomio de Lagrange para los siguientes datos y grafique:

- $(0,0), \ (30,0.5), \left(60,\dfrac{\sqrt{3}}{2}\right), \ (90,1)$


$$
x_0 = 0,\quad x_1 = 30,\quad x_2 = 60,\quad x_3 = 90
$$

$$
y_0 = 0,\quad y_1 = 0.5,\quad y_2 = \tfrac{\sqrt{3}}{2},\quad y_3 = 1
$$

Polinomio 

$$
P(x)=
0\,L_0(x)
+\frac{1}{2}L_1(x)
+\frac{\sqrt{3}}{2}L_2(x)
+1\cdot L_3(x)
$$

Bases de Lagrange

$$
L_0(x)=\frac{(x-30)(x-60)(x-90)}
{(0-30)(0-60)(0-90)}
=\frac{(x-30)(x-60)(x-90)}{162000}
$$



$$
L_1(x)=\frac{x(x-60)(x-90)}
{(30-0)(30-60)(30-90)}
=\frac{x(x-60)(x-90)}{54000}
$$


$$
L_2(x)=\frac{x(x-30)(x-90)}
{(60-0)(60-30)(60-90)}
=-\frac{x(x-30)(x-90)}{54000}
$$



$$
L_3(x)=\frac{x(x-30)(x-60)}
{(90-0)(90-30)(90-60)}
=\frac{x(x-30)(x-60)}{162000}
$$


$$
P(x)=
\frac{1}{2}\left(\frac{x(x-60)(x-90)}{54000}\right)
+
\frac{\sqrt{3}}{2}\left(\frac{x(x-30)(x-90)}{54000}\right)
+
\left(\frac{x(x-30)(x-60)}{162000}\right)
$$



$$
P(x)=
-6.054\times 10^{-7}x^{3}
-1.9943\times 10^{-5}x^{2}
+0.0178098x
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import lagrange

x_pts = np.array([0.0, 30.0, 60.0, 90.0])
y_pts = np.array([0.0, 0.5, np.sqrt(3)/2, 1.0])

P = lagrange(x_pts, y_pts)

x_min, x_max = -10, 100
y_min, y_max = -0.5, 1.5

x_vals = np.linspace(x_min, x_max, 800)
y_vals = P(x_vals)

plt.figure(figsize=(9, 5))
plt.plot(x_vals, y_vals, label='Polinomio de Lagrange', color='red', linewidth=2)
plt.scatter(x_pts, y_pts, color='green', label='Puntos dados', zorder=5, s=60)

plt.xlabel("x (grados)")
plt.ylabel("y")
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.title("Interpolación de Lagrange para los puntos (0,0), (30,0.5), (60,√3/2),(90,1)")
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

- $(1,1), \ (2,2), \ (3,2)$


$$
x_0 = 1,\quad x_1 = 2,\quad x_2 = 3
$$

$$
y_0 = 1,\quad y_1 = 2,\quad y_2 = 2
$$


Polinomio 

$$
P(x)=
1\cdot L_0(x)
+2\cdot L_1(x)
+2\cdot L_2(x)
$$

Bases de Lagrange


$$
L_0(x)=\frac{(x-2)(x-3)}{(1-2)(1-3)}
      =\frac{(x-2)(x-3)}{2}
$$


$$
L_1(x)=\frac{(x-1)(x-3)}{(2-1)(2-3)}
      =-\,(x-1)(x-3)
$$


$$
L_2(x)=\frac{(x-1)(x-2)}{(3-1)(3-2)}
      =\frac{(x-1)(x-2)}{2}
$$

$$
P(x)=1\left(\frac{(x-2)(x-3)}{2}\right)
-2\left((x-1)(x-3)\right)
+2\left(\frac{(x-1)(x-2)}{2}\right)
$$

$$
P(x)= -\frac{1}{2}x^2+\frac{5}{2}x-1
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import lagrange

x_pts = np.array([1.0, 2.0, 3.0])
y_pts = np.array([1.0, 2.0, 2.0])

P = lagrange(x_pts, y_pts)

x_min, x_max = 0, 4
y_min, y_max = 0, 3

x_vals = np.linspace(x_min, x_max, 400)
y_vals = P(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, label='Polinomio de Lagrange', color='red', linewidth=2)
plt.scatter(x_pts, y_pts, color='green', label='Puntos dados', zorder=5, s=60)

plt.title("Interpolación de Lagrange para los puntos (1,1), (2,2), (3,2)")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

- $(-2,5), \ (1,7), \ (3,11), \ (7,34)$

$$
x_0=-2,\quad x_1=1,\quad x_2=3,\quad x_3=7

$$
y_0=5,\quad y_1=7,\quad y_2=11,\quad y_3=34
$$


$$
P(x) = 5\,L_0(x) + 7\,L_1(x) + 11\,L_2(x) + 34\,L_3(x)

Bases de Lagrange



$$
L_0(x)=\frac{(x-1)(x-3)(x-7)}{(-2-1)(-2-3)(-2-7)}
=\frac{(x-1)(x-3)(x-7)}{-135}
$$



$$
L_1(x)=\frac{(x+2)(x-3)(x-7)}{(1+2)(1-3)(1-7)}
=\frac{(x+2)(x-3)(x-7)}{36}
$$


$$
L_2(x)=\frac{(x+2)(x-1)(x-7)}{(3+2)(3-1)(3-7)}
=\frac{(x+2)(x-1)(x-7)}{40}
$$


$$
L_3(x)=\frac{(x+2)(x-1)(x-3)}{(7+2)(7-1)(7-3)}
=\frac{(x+2)(x-1)(x-3)}{216}
$$


$$
P(x) = 5\cdot\frac{(x-1)(x-3)(x-7)}{-135} \;+\; 7\cdot\frac{(x+2)(x-3)(x-7)}{36} \;+\; 11\cdot\frac{(x+2)(x-1)(x-7)}{-40} \;+\; 34\cdot\frac{(x+2)(x-1)(x-3)}{216}
$$

$$P(x) = -\frac{1}{27}\,(x-1)(x-3)(x-7) \;+\; \frac{7}{36}\,(x+2)(x-3)(x-7) \;-\; \frac{11}{40}\,(x+2)(x-1)(x-7) \;+\; \frac{17}{108}\,(x+2)(x-1)(x-3)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import lagrange

x_pts = np.array([-2, 1, 3, 7])
y_pts = np.array([5, 7, 11, 34])

P = lagrange(x_pts, y_pts)

x_min, x_max = -5, 14
y_min, y_max = 0, 50

x_vals = np.linspace(x_min, x_max, 500)
y_vals = P(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, label='Polinomio de Lagrange', color='red', linewidth=2)
plt.scatter(x_pts, y_pts, color='green', label='Puntos dados', zorder=5)

plt.title("Interpolación de Lagrange para los puntos (-2,5), (1,7), (3,11), (7,34) ")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
